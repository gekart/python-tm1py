Python crams a lot of meaning into a handful of symbols, often with the same character meaning very different things in different contexts. Here's a thorough sweep, grouped so the overloaded characters are easy to compare.

## Arithmetic and numeric

| Symbol           | Meaning                                                                      |
| ---------------- | ---------------------------------------------------------------------------- |
| `+ - * /`        | Add, subtract, multiply, true division (always returns float).               |
| `//`             | Floor division. `7 // 2 == 3`, `-7 // 2 == -4`.                              |
| `%`              | Modulo for numbers; old-style string formatting (`"hi %s" % name`).          |
| `**`             | Exponentiation. `2 ** 10`. Also `pow(a, b, mod)` has no operator equivalent. |
| `@`              | Matrix multiplication (`__matmul__`). Used by NumPy: `A @ B`.                |
| `1_000_000`      | Underscores as digit separators in numeric literals.                         |
| `0b`, `0o`, `0x` | Binary, octal, hex literal prefixes.                                         |
| `1j`, `2+3j`     | Complex number literals.                                                     |
| `1e6`, `1.5e-3`  | Scientific notation.                                                         |

## Comparisons and identity

| Symbol         | Meaning                                                                          |
| -------------- | -------------------------------------------------------------------------------- |
| `== !=`        | Value equality / inequality (calls `__eq__`).                                    |
| `< > <= >=`    | Ordering.                                                                        |
| `is`, `is not` | Identity — same object in memory. Use for `None`, never for value comparison.    |
| `in`, `not in` | Membership. Calls `__contains__`.                                                |
| `1 < x < 10`   | Chained comparison. Equivalent to `1 < x and x < 10`, but `x` is evaluated once. |

## Logical

| Symbol             | Meaning                                                                                               |
| ------------------ | ----------------------------------------------------------------------------------------------------- |
| `and`, `or`, `not` | Short-circuit logical operators. `and`/`or` return one of the operands, not a bool: `x or "default"`. |
| `x if cond else y` | Conditional (ternary) expression.                                                                     |

## Bitwise — and their second lives

| Symbol     | Numeric meaning | Other meanings                                                                      |
| ---------- | --------------- | ----------------------------------------------------------------------------------- |
| `&`        | Bitwise AND     | Set intersection: `{1,2} & {2,3}`. Pandas/NumPy element-wise AND on boolean arrays. |
| `\|`       | Bitwise OR      | Set union. Dict merge (3.9+): `d1 \| d2`. Type union (3.10+): `int \| None`.        |
| `^`        | Bitwise XOR     | Set symmetric difference.                                                           |
| `~`        | Bitwise NOT     | Boolean negation in NumPy/pandas: `df[~mask]`.                                      |
| `<<`, `>>` | Bit shifts      | —                                                                                   |

## Assignment

| Symbol                                        | Meaning                                                                                |
| --------------------------------------------- | -------------------------------------------------------------------------------------- |
| `=`                                           | Bind name to value. **Not** an expression in normal contexts.                          |
| `+= -= *= /= //= %= **= &= \|= ^= <<= >>= @=` | Augmented assignment. For mutable objects these may mutate in place (`__iadd__` etc.). |
| `:=`                                          | Walrus / assignment expression (3.8+). Assigns and returns: `if (n := len(x)) > 10:`.  |
| `a = b = c = 0`                               | Chained assignment — same value bound to all names.                                    |
| `a, b = b, a`                                 | Tuple swap (really tuple packing + unpacking).                                         |

## The star and double-star — the most overloaded characters

Both `*` and `**` mean wildly different things depending on where they appear.

`*` in expressions:

- **Multiplication / repetition**: `3 * 4`, `[0] * 10`, `"-" * 20`.
- **Iterable unpacking** in calls: `f(*lst)` spreads list as positional args.
- **Iterable unpacking** in literals (3.5+): `[*a, *b]`, `(*a, 0)`, `{*a, *b}`.
- **Extended unpacking** in assignment: `first, *middle, last = seq`. The starred name catches everything else as a list.
- **Bare `*` in function signature**: marks the boundary after which all parameters are keyword-only. `def f(a, *, b):` — `b` must be passed by name.
- **`*args`**: collects extra positional arguments into a tuple.

`**` in expressions:

- **Exponentiation**: `2 ** 8`.
- **Dict unpacking** in calls: `f(**d)`.
- **Dict unpacking** in literals (3.5+): `{**a, **b, "x": 1}`.
- **`**kwargs`\*\*: collects extra keyword arguments into a dict.

## Function signature special markers

```
def f(pos1, pos2, /, normal, *, kw_only1, kw_only2):
    ...
```

| Symbol              | Meaning                                                                      |
| ------------------- | ---------------------------------------------------------------------------- |
| `/` (alone, 3.8+)   | Everything **before** is positional-only — cannot be passed by name.         |
| `*` (alone)         | Everything **after** is keyword-only.                                        |
| `*args`             | Catch remaining positional args as tuple.                                    |
| `**kwargs`          | Catch remaining keyword args as dict.                                        |
| `name: T`           | Parameter type annotation.                                                   |
| `name: T = default` | Annotation with default.                                                     |
| `-> T`              | Return type annotation.                                                      |
| `def f(x=[]):`      | Mutable default arg — classic gotcha; default is evaluated once at def time. |

## Slicing

`seq[start:stop:step]` — any of the three can be omitted.

| Idiom             | Meaning                         |
| ----------------- | ------------------------------- |
| `a[:]`            | Shallow copy (for lists, etc.). |
| `a[::-1]`         | Reversed copy.                  |
| `a[::2]`          | Every second element.           |
| `a[1:]`, `a[:-1]` | Drop first / drop last.         |
| `a[i:i] = [x, y]` | Insert without removing.        |
| `del a[i:j]`      | Delete slice.                   |

`x[i, j]` is a tuple index — `__getitem__((i, j))`. NumPy/pandas use this; on plain dicts it just becomes a tuple key. NumPy also accepts `...` (Ellipsis) inside indexing as "all remaining axes": `arr[..., 0]`.

## Type annotation syntax

| Symbol                                                  | Meaning                                                                                  |
| ------------------------------------------------------- | ---------------------------------------------------------------------------------------- |
| `x: int = 0`                                            | Variable annotation.                                                                     |
| `-> T`                                                  | Return type.                                                                             |
| `list[int]`, `dict[str, int]`                           | Generic subscription (3.9+). Pre-3.9: `List[int]` from `typing`.                         |
| `int \| None`                                           | Union type (3.10+). Equivalent to `Optional[int]` / `Union[int, None]`.                  |
| `Callable[[int], str]`                                  | Function type.                                                                           |
| `...` (Ellipsis)                                        | Used in stubs (`def f() -> int: ...`) and types (`Callable[..., T]`, `tuple[int, ...]`). |
| `TypeVar`, `Generic[T]`, `T` in `class Box[T]:` (3.12+) | Generics.                                                                                |

## Decorators

```
@cache
@dataclass(frozen=True)
def f(): ...
```

Just syntactic sugar: `f = cache(dataclass(frozen=True)(f))`. Common ones: `@property`, `@classmethod`, `@staticmethod`, `@functools.wraps`.

## String literal prefixes

Any combination, in either case:

| Prefix                     | Meaning                                                                                                        |
| -------------------------- | -------------------------------------------------------------------------------------------------------------- |
| `"..."`, `'...'`           | Plain `str`.                                                                                                   |
| `"""..."""`                | Triple-quoted; spans lines.                                                                                    |
| `r"..."`                   | Raw — backslashes are literal. Common for regex and Windows paths.                                             |
| `b"..."`                   | `bytes` literal.                                                                                               |
| `f"..."`                   | f-string. `f"{x=}"` self-documents (3.8+); `f"{x:.2f}"` format spec.                                           |
| `rb"..."`, `fr"..."`, etc. | Combine prefixes. `bf""` is **not** allowed.                                                                   |
| `u"..."`                   | Unicode (no-op in Py3, kept for 2/3 compat).                                                                   |
| `"foo" "bar"`              | Adjacent string literals concatenate at compile time → `"foobar"`. Useful for long strings split across lines. |

## Collection literals

| Syntax                        | Meaning                                                                |
| ----------------------------- | ---------------------------------------------------------------------- |
| `[...]`                       | List.                                                                  |
| `(...)`                       | Tuple — but `(x)` is just `x`; tuple needs comma: `(x,)` or just `x,`. |
| `{...}`                       | Set or dict — `{}` is an **empty dict**, empty set is `set()`.         |
| `{1, 2, 3}` vs `{1: 2}`       | Set vs dict.                                                           |
| `[x for x in xs]`             | List comprehension.                                                    |
| `{x for x in xs}`             | Set comprehension.                                                     |
| `{k: v for ...}`              | Dict comprehension.                                                    |
| `(x for x in xs)`             | Generator expression — lazy, single-pass.                              |
| `sum(x*x for x in xs)`        | Bare gen-exp inside a call — parens optional.                          |
| `[x for x in xs if cond]`     | Filtering clause.                                                      |
| `[x for xs in m for x in xs]` | Nested loops; outer to inner, left to right.                           |

## The underscore — many lives

| Form                | Meaning                                                                    |
| ------------------- | -------------------------------------------------------------------------- |
| `_`                 | Conventionally "throwaway" variable: `for _ in range(10):`.                |
| `_` (in REPL)       | Last evaluated expression's value.                                         |
| `_name`             | Convention: "private" / internal use.                                      |
| `__name`            | Name-mangled inside a class to `_ClassName__name`.                         |
| `__name__`          | Dunder ("magic") attribute or method — defined by Python.                  |
| `_` (in `match`)    | Wildcard pattern (3.10+).                                                  |
| `_` (function name) | Conventional alias for `gettext.gettext` in i18n.                          |
| `1_000_000`         | Digit separator in numeric literal.                                        |
| `_, *_, x` etc.     | Multiple `_` allowed in unpacking; the values are just thrown into a list. |

Notable dunders include `__init__`, `__repr__`, `__str__`, `__len__`, `__iter__`, `__next__`, `__getitem__`, `__call__`, `__enter__`/`__exit__`, `__eq__`/`__hash__`, `__add__` and friends.

## Ellipsis `...`

A real singleton object (`Ellipsis`). Uses:

- Placeholder body: `def f(): ...` (equivalent to `pass`).
- Stub files (`.pyi`) for type info.
- NumPy multi-axis indexing: `arr[..., 0]`.
- Type expressions: `Callable[..., int]`, `tuple[int, ...]`.

## Control-flow / definition keywords that look like operators

| Keyword                   | Meaning                                                                                                                                  |
| ------------------------- | ---------------------------------------------------------------------------------------------------------------------------------------- |
| `lambda x, y: x+y`        | Anonymous single-expression function.                                                                                                    |
| `yield x`                 | Pause function, return value, resume on next call → makes a generator.                                                                   |
| `yield from iterable`     | Delegate to a sub-iterator (3.3+).                                                                                                       |
| `async def`, `await`      | Coroutines (3.5+).                                                                                                                       |
| `async for`, `async with` | Async iteration / context.                                                                                                               |
| `global x`, `nonlocal x`  | Rebind name in outer scope.                                                                                                              |
| `match x: case ...`       | Structural pattern matching (3.10+). Patterns include `_`, `\|` (alternatives), `as`, capture binding, class patterns `Point(x=0, y=_)`. |

## Punctuation that's easy to miss

| Symbol                     | Meaning                                                                            |
| -------------------------- | ---------------------------------------------------------------------------------- |
| `:`                        | Block introducer (`if`, `for`, `def`, `class`); slice; dict pair; type annotation. |
| `,`                        | Tuple builder; argument separator. Trailing comma is allowed almost everywhere.    |
| `;`                        | Statement separator on one line — legal but discouraged.                           |
| `\` at line end            | Explicit line continuation. Usually unnecessary inside `()`, `[]`, `{}`.           |
| `#`                        | Comment to end of line.                                                            |
| `# type: ...`              | Legacy type-comment syntax (rare now).                                             |
| `# noqa`, `# type: ignore` | Tooling pragmas, not Python syntax.                                                |

## A few idioms worth knowing as units

```
if __name__ == "__main__":      # script entry point
    main()

x, y = y, x                      # swap

a, *rest = seq                   # head/tail split

result = expensive() if (cached := lookup()) is None else cached  # walrus

with open(p) as f, lock:         # multiple context managers
    ...

{**defaults, **overrides}        # merge dicts (or `defaults | overrides` 3.9+)

[*a, *b]                         # concatenate iterables into list

sorted(items, key=lambda x: (x.priority, -x.age))  # tuple key for multi-sort

text = (
    "line one "
    "line two"                   # adjacent-string concat, no `+` cost
)
```
