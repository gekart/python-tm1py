# Pandas Methods — Grouped Reference

Methods marked with ◆ are lesser-known but worth knowing.
Methods marked with ⚡ are performance-relevant.
`df` = DataFrame, `s` = Series.

---

## 1. I/O — Reading

### Common

- `pd.read_csv` — flat files; key args: `dtype`, `parse_dates`, `usecols`, `chunksize`, `nrows`, `na_values`, `converters`
- `pd.read_excel` — `.xlsx`/`.xls`; supports `sheet_name=None` for all sheets as dict
- `pd.read_json` — JSON; pair with `pd.json_normalize` for nested
- `pd.read_parquet` ⚡ — columnar, fast, typed
- `pd.read_sql` / `read_sql_query` / `read_sql_table` — DB
- `pd.read_html` ◆ — pulls all `<table>` elements from a URL/HTML
- `pd.read_clipboard` ◆ — paste tabular data straight from OS clipboard

### Less common

- `pd.read_feather` ⚡ ◆ — Arrow-based, very fast
- `pd.read_orc` ◆
- `pd.read_hdf` ◆ — HDF5 store, supports queries
- `pd.read_pickle` ◆ — round-trip Python objects (untrusted-source warning)
- `pd.read_fwf` ◆ — fixed-width files
- `pd.read_xml` ◆
- `pd.read_sas` / `read_spss` / `read_stata` ◆ — stats software formats
- `pd.read_table` — like `read_csv` but tab-default

### Helpers

- `pd.json_normalize` ◆ — flatten nested JSON/dicts to a flat DataFrame
- `pd.ExcelFile` ◆ — open once, read multiple sheets without re-parsing

## 4. Selection & Indexing

### Standard

- `df[col]`, `df[[cols]]`, `df[bool_mask]`
- `df.loc[row_label, col_label]` — label-based, end-inclusive
- `df.iloc[row_pos, col_pos]` — position-based, end-exclusive
- `df.at[r, c]` ⚡ / `df.iat[r, c]` ⚡ — single-cell, fastest
- `df.get(key, default=None)` — won't raise on missing
- `df.take(indices, axis=)` — reorder/select by integer position

### Less known

- `df.xs(key, level=, axis=)` ◆ — cross-section on a MultiIndex
- `df.filter(items=, like=, regex=, axis=)` ◆ — pick labels by pattern (NOT row filtering)
- `df.truncate(before=, after=)` ◆ — slice by label range, both endpoints inclusive
- `df.query("col > 5 and other == 'a'")` ◆ — readable filtering
- `df.eval("c = a + b")` ⚡ ◆ — string-expression evaluation, can be faster on large frames
- `df.lookup` — removed in 2.0 (use `melt`/`stack` patterns)
- `s.searchsorted(values)` ◆ — bisect on a sorted Series
- `df.idxmin`, `df.idxmax` — label of min/max (vs `argmin/argmax` for position)

## 7. Transformation — Apply Family

- `s.map(func_or_dict)` — element-wise on Series
- `df.map(func)` — element-wise on DataFrame (was `applymap`, renamed in 2.1)
- `df.apply(func, axis=)` — per-row or per-column
- `gb.transform(func)` — broadcast result back to original shape
- `df.agg(func_or_list_or_dict)` — aggregate; can mix per-col aggs
- `df.pipe(func, *args, **kwargs)` ◆ — chain custom functions cleanly
- `df.assign(new_col=lambda d: ...)` ◆ — chainable column creation
- `df.eval("c = a*b")` ⚡ ◆

**Idiom worth knowing:**

```
(df
 .pipe(load_data)
 .assign(ratio=lambda d: d.a / d.b)
 .query("ratio > 0.5")
 .groupby("group")
 .agg(mean_r=("ratio", "mean")))
```

## 10. GroupBy

### Standard aggregations

`count`, `sum`, `mean`, `median`, `std`, `var`, `sem`, `min`, `max`, `prod`, `first`, `last`, `size`

### Less known on GroupBy

- `gb.nth(n)` ◆ — nth row of each group (supports lists, slices)
- `gb.ngroup()` ◆ — integer group label per row
- `gb.cumcount()` ◆ — within-group row number
- `gb.get_group(key)` ◆ — extract one group as a DataFrame
- `gb.indices` / `gb.groups` ◆ — peek at index mapping
- `gb.head(n)` / `gb.tail(n)` — first/last n per group
- `gb.filter(lambda g: ...)` ◆ — drop entire groups by predicate
- `gb.transform(...)` — same shape back
- `gb.apply(...)` — most flexible, slowest
- `gb.pipe(...)` ◆
- `gb.rolling(...)`, `gb.expanding(...)`, `gb.ewm(...)` ◆ — windows within groups
- `gb.ffill()` / `gb.bfill()` ◆ — fill within groups, not across

### Named aggregations

```
df.groupby("k").agg(
    avg_x=("x", "mean"),
    n_unique=("y", "nunique"),
    custom=("z", lambda s: s.max() - s.min()),
)
```

- `df.expanding(min_periods=)` — cumulative window

## 17. Iteration (use sparingly)

- `df.iterrows()` — slow, returns (index, Series)

## 21. Performance cheatsheet ⚡

- Read **parquet/feather** instead of CSV when possible.
- Use **categorical** dtype for low-cardinality strings.
- Use **nullable dtypes** (`Int64`, `Float64`, `string`, `boolean`) when you have NaNs and want correct types.
- Prefer **vectorized ops** > `apply` > `iterrows`.
- For arithmetic on big frames: `df.eval` / `pd.eval` / numexpr.
- `df.itertuples()` ≫ `df.iterrows()` if you must loop.
- `at`/`iat` for single scalar access; `loc`/`iloc` for slicing.
- `df.nlargest(n, col)` instead of `sort_values(col).head(n)`.
- Avoid chained assignment (`df[df.x>0]['y'] = ...`); use `.loc`.
- For huge data: read in `chunksize=` chunks, or switch to **PyArrow** backend (`dtype_backend='pyarrow'`).
