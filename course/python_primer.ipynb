## 1. Python in the language landscape

Python was designed by Guido van Rossum in 1989 and has been one of the
most widely used languages in the world for the better part of two
decades. It is interpreted (no compile step), dynamically typed (a
variable's type is a property of the value it holds, not of the
variable itself), garbage collected, and shipped with a substantial
standard library. The reference implementation, CPython, is the one
almost everyone uses; the others (PyPy, Jython, IronPython) exist for
specific niches.

Three properties shape almost everything that follows. First, **Python
values are objects**, including small integers and short strings; there
is no primitive vs. object distinction the way Java has between `int`
and `Integer`. Second, **indentation is syntax**, not formatting. A
block of code is defined by being indented one level deeper than its
header, and there are no curly braces. Third, **the standard library is
large and well documented**. A surprising amount of work that would
require a third party dependency in other languages is one
`import` away in Python.

For the planning analytics work this course is aimed at, Python is the
glue between TM1, files, databases, and reports. The same language
reads a CSV, calls a TM1 cube, transforms a DataFrame, and produces a
chart. There are deeper languages and faster languages, but no other
mainstream language has the same combination of "write a small script
on Tuesday afternoon" and "scale that to a production pipeline by
Friday".

## 2. Running Python

Python code runs in three places. The choice between them is a matter
of feedback loop, not capability.

The **interpreter** (the REPL, short for read evaluate print loop)
opens by typing `python` (or `python3` on systems where Python 2 is
still around) at a shell prompt. It reads one expression at a time,
prints the result, and waits for the next.

```text
$ python
Python 3.12.0 (main, Oct  2 2023, ...)
>>> 2 + 2
4
>>> "hello"
'hello'
>>> exit()
```

A **script** is a plain text file ending in `.py`. The same code that
works in the REPL works in a script, except that printed output is no
longer automatic; only `print(...)` calls produce output. A script is
run with `python script.py` and its execution is top to bottom.

````
# sales_summary.py
total = 120_000 + 250_000 + 90_000
print("Total revenue:", total)
````

A **notebook** (Jupyter, JupyterLab, VS Code's notebook view) is a file
of cells, each cell a small block of code that runs in the same
interpreter session as the cells before it. Notebooks behave like a
REPL whose history is saved and editable. They are the standard format
for exploratory data work and the format the exercises in this course
use. The last expression in a cell is auto displayed without a
`print(...)` call, which is a notebook convention rather than a Python
one.

Most readers will spend their first weeks in a notebook, then graduate
to scripts as work moves into production. Both are first class; pick
the one that matches the task.

## 3. Hello world and the print function

The traditional first program prints a fixed message and exits. In
Python the whole program is one line.

````
print("Hello, world!")
# Hello, world!
````

`print` is a built in function. It accepts any number of arguments and
writes them to standard output, separated by spaces and followed by a
newline. The defaults are configurable through keyword arguments:

````
print("North", "South", "East", "West")
# North South East West

print("North", "South", "East", "West", sep=", ")
# North, South, East, West

print("loading", end="")
print("...done")
# loading...done

print("error: file missing", file=sys.stderr)   # routed to stderr instead of stdout
````

Two facts are worth absorbing now. First, `print` always converts its
arguments to strings (calling `str(value)` on each), so mixing types is
fine: `print("count:", 42, "items")` works without any explicit
conversion. Second, the parentheses are mandatory; `print` is a
function, not a statement. The Python 2 form (`print "hello"` without
parentheses) was removed in Python 3 and is now a syntax error.

For anything beyond the simplest "print this value" call, prefer an f
string (Topic 6) over multiple comma separated arguments. The result
reads more like a sentence and is easier to format precisely.

## 4. Variables and assignment

A Python variable is a name bound to a value. The assignment statement
creates the binding; the value already exists as an object somewhere in
memory, and the name simply points at it.

````
revenue = 120_000
region  = "Europe"
year    = 2026
````

There is no declaration; the first assignment to a name creates it,
and a later assignment to the same name rebinds it (possibly to a
value of a different type, since types belong to values, not to
variables).

````
sales = 100
sales = sales + 50         # rebinds to 150
sales = "growing"          # rebinds to a string; legal, often a bug
````

Several patterns make assignment more compact and idiomatic:

````
a, b = 1, 2                # tuple unpacking: a = 1, b = 2
a, b = b, a                # swap, no temporary variable
x = y = 0                  # chained assignment: both names bound to the same 0

revenue += 10_000          # augmented assignment, equivalent to revenue = revenue + 10_000
````

Naming conventions matter for readability. The convention is
`snake_case` for variables and functions, `PascalCase` for classes,
`UPPER_SNAKE_CASE` for module level constants, and a leading underscore
(`_helper`) for names that are private to the module. The underscore
itself (`_`) is used as a "throwaway" name when a value is needed
syntactically but not semantically (`for _ in range(3): ...`).

A name can hold any value, including a function, a class, or a module.
This is what makes patterns like callbacks and decorators natural in
Python; functions are not a special category, they are objects like
the rest.

## 5. Numbers and arithmetic

Python has two numeric types in everyday use. `int` is an arbitrary
precision integer; `float` is a 64 bit IEEE 754 floating point number.
The literal `42` is an `int`; the literal `42.0` is a `float`. Mixing
the two in arithmetic produces a `float`.

````
revenue = 120_000           # int (the underscores are visual separators)
margin_rate = 0.30          # float
margin = revenue * margin_rate
print(margin)               # 36000.0
````

The arithmetic operators are the conventional ones, with one trap and
one convenience.

````
2 + 3      # 5
2 - 3      # -1
2 * 3      # 6
2 ** 3     # 8     exponentiation
7 / 2      # 3.5   true division, always returns float
7 // 2     # 3     floor division, returns int when both operands are int
7 % 2      # 1     modulo
abs(-3)    # 3
round(3.7) # 4
````

The trap: `/` is **true division** and always returns a float. To get
the integer quotient, use `//`. This is a deliberate Python 3 change
from Python 2, where `/` between two integers produced an integer; a
lot of older code on the internet still has it the other way.

The convenience: integers are arbitrary precision. There is no
overflow, and no need for `long` or `BigInteger` types. `2 ** 100` is
just an int.

Floating point arithmetic has the usual binary representation
surprises (`0.1 + 0.2 == 0.3` is `False`); Python is not unusual in
this regard. For currency or any other domain where exact decimal
arithmetic matters, the standard library `decimal` module provides
`Decimal`, which behaves like the calculator on the desk:

````
from decimal import Decimal

price = Decimal("19.99")
units = 3
total = price * units       # Decimal('59.97')
````

Initialise `Decimal` from a string, not from a float; `Decimal(0.1)`
captures the binary approximation rather than the value the author
meant.

## 6. Strings and f strings

A string is an immutable sequence of Unicode characters. Either single
or double quotes work; the choice is style. Triple quoted strings span
multiple lines.

````
region = "Europe"
note   = 'first quarter'
multiline = """First line
second line
third line"""
````

Strings concatenate with `+` and repeat with `*`:

````
"North" + " " + "America"   # 'North America'
"-" * 20                    # '--------------------'
````

Indexing is zero based; slicing uses the half open `start:stop:step`
syntax familiar from many languages.

````
text = "Sales Plan 2026"
text[0]      # 'S'
text[-1]     # '6'    negative indices count from the end
text[0:5]    # 'Sales'
text[6:]     # 'Plan 2026'
text[::2]    # 'SlsPa 06'   every second character
````

Strings are immutable. Methods that look like they would mutate (`upper`,
`replace`, `strip`) all return a new string and leave the original
alone:

````
"  hello  ".strip()              # 'hello'
"hello".upper()                  # 'HELLO'
"Sales Plan".replace("Plan", "Forecast")   # 'Sales Forecast'
"a,b,c".split(",")               # ['a', 'b', 'c']
",".join(["a", "b", "c"])        # 'a,b,c'
"sales".startswith("sa")         # True
len("Sales")                     # 5
````

The most useful single feature for printing values is the **f string**:
a string literal prefixed with `f` whose `{...}` placeholders are
evaluated and inserted at runtime.

````
region = "Europe"
revenue = 123_456.78

print(f"{region}: {revenue:,.2f}")
# Europe: 123,456.78

print(f"share: {revenue / 1_000_000:.1%}")
# share: 12.3%
````

The text after the colon is a format specifier: `,.2f` means "comma
thousands separator, two decimal places, fixed point", `.1%` means
"one decimal place, percent", `>10` means "right align in a field of
width 10". The full format mini language is in the standard library
documentation; the patterns above cover most everyday formatting.

## 7. Booleans, None, and truthiness

The boolean type has exactly two values, `True` and `False`. They are
the result of comparisons (`==`, `<`, `>=`, `in`) and combine with the
keywords `and`, `or`, `not`.

````
x = 5
x > 0              # True
x > 0 and x < 10   # True
not (x == 0)       # True
````

`None` is a separate singleton value used to represent "nothing here".
It is the conventional return of a function that does not explicitly
return anything, and it is the right initial value for a variable that
will be set later.

````
last_error = None
result     = None
````

Test for `None` with the `is` operator, not `==`: `if x is None` is
the convention and the tools that lint Python code expect it.

Python applies a generalisation called **truthiness** to `if`, `while`,
and the boolean operators: any value can stand in for a boolean. The
"falsy" values are `False`, `None`, `0`, `0.0`, the empty string `""`,
and the empty container `[]`, `{}`, `()`, `set()`. Everything else is
truthy.

````
items = []
if items:
    print("got items")
else:
    print("nothing")
# nothing
````

This makes idiomatic Python noticeably terser than the equivalent in
Java or C#. The trade off is that `if x:` and `if x is not None:`
behave differently when `x` is `0` or `""`. When the distinction
matters (a value that may be missing vs. a value that may legitimately
be zero), use the explicit form.

The boolean operators short circuit. `a and b` evaluates `b` only if
`a` is truthy; `a or b` evaluates `b` only if `a` is falsy. They also
return one of the operands, not necessarily a `True` / `False` value:
`"" or "default"` returns `"default"`. This is the canonical Python
form for "use this value, or fall back to that one".

## 8. Lists and tuples

A list is an ordered, mutable, heterogeneous sequence. A tuple is an
ordered, immutable, heterogeneous sequence. The two share most of
their interface; the difference is whether the contents can change
after construction.

````
regions: list[str] = ["Europe", "Americas", "Asia", "Pacific"]
regions[0]                    # 'Europe'
regions[-1]                   # 'Pacific'
regions[1:3]                  # ['Americas', 'Asia']
len(regions)                  # 4

regions.append("Africa")      # mutates: ['Europe', ..., 'Africa']
regions.extend(["Antarctica"])
regions.insert(0, "HQ")
regions.pop()                 # removes and returns the last element
regions.remove("HQ")          # removes the first occurrence by value
````

Tuples use parentheses (or no brackets at all if the context is clear)
and support indexing and slicing but not mutation:

````
period: tuple[int, str] = (2026, "Q1")
year, quarter = period             # tuple unpacking
print(year)                        # 2026
print(quarter)                     # 'Q1'

# period[0] = 2027                 # TypeError: tuples are immutable
````

A common rule of thumb: **lists for collections of similar things,
tuples for fixed shape records**. A list of regions can grow and
shrink; a `(year, quarter)` pair always has exactly two slots with
specific meanings. The distinction is convention rather than language
enforcement, but it pays off in readability.

The `in` operator tests membership; both lists and tuples support it,
along with iteration:

````
"Europe" in regions       # True
"Mars"   in regions       # False

for region in regions:
    print(region)
````

Lists support a few operations that are easy to miss but heavily used:
`sorted(regions)` returns a new sorted list, `regions.sort()` sorts in
place, `list(reversed(regions))` reverses, and `regions.index("Asia")`
returns the position of the first match (raising `ValueError` if not
present).

## 9. Dictionaries and sets

A dict (also called a hash map, associative array, or object literal in
other languages) maps keys to values. Keys must be hashable (immutable
types like strings, numbers, and tuples qualify; lists and dicts do
not). Values can be anything.

````
revenue: dict[str, float] = {
    "Europe":   120_000.0,
    "Americas": 250_000.0,
    "Asia":      90_000.0,
    "Pacific":   60_000.0,
}

revenue["Europe"]              # 120000.0
revenue["Africa"] = 30_000.0   # adds a new key
revenue["Europe"] += 5_000.0   # mutates an existing value

"Asia" in revenue              # True
revenue.get("Mars", 0.0)       # 0.0; .get returns the default if the key is missing
del revenue["Africa"]          # removes the key
len(revenue)                   # 4
````

Iteration over a dict yields its keys; `.values()` yields the values,
`.items()` yields `(key, value)` pairs. Since Python 3.7, dicts
preserve insertion order; the first key inserted is the first key
yielded.

````
for region in revenue:
    print(region, revenue[region])

for region, amount in revenue.items():
    print(f"{region}: {amount:,.0f}")
````

A set is an unordered collection of unique, hashable values. The
typical use is membership testing and deduplication.

````
seen: set[str] = set()
seen.add("Europe")
seen.add("Asia")
seen.add("Europe")           # duplicate; ignored
len(seen)                    # 2

regions_a = {"Europe", "Asia", "Pacific"}
regions_b = {"Asia", "Pacific", "Americas"}
regions_a & regions_b        # {'Asia', 'Pacific'} intersection
regions_a | regions_b        # union
regions_a - regions_b        # difference
````

Sets are not the right tool when ordering or counts matter; they are
the right tool when the question is "is this in the collection?" or
"what are the unique values?". A `set` membership test runs in
constant time, while `value in some_list` scales with the length of
the list.

## 10. The if statement

The `if` statement runs a block of code when a condition is truthy.
Optional `elif` and `else` branches handle alternatives. Indentation
defines the block; the colon at the end of the header is mandatory.

````
revenue = 123_000

if revenue >= 200_000:
    rating = "high"
elif revenue >= 100_000:
    rating = "medium"
else:
    rating = "low"

print(rating)               # 'medium'
````

The comparison operators are the conventional ones (`==`, `!=`, `<`,
`<=`, `>`, `>=`) plus two Python specific membership tests (`in`,
`not in`) and identity tests (`is`, `is not`). One Python convenience
is **chained comparisons**: `a < b < c` means `a < b and b < c`,
evaluating `b` only once.

````
x = 50
if 0 < x < 100:
    print("in range")
````

For small conditional value choices, the **conditional expression**
(also called the ternary form) reads more compactly than a four line
`if`/`else`:

````
status = "ok" if revenue > 0 else "missing"
````

The form is `value_if_true if condition else value_if_false`. It is a
matter of taste when to reach for it; a useful guideline is to use it
when the entire expression fits on one line and the meaning is
immediately readable.

Three idioms that newcomers sometimes miss. First, there is no `switch`
statement in Python before 3.10; an `if`/`elif` chain is the
conventional substitute, and a dict from value to outcome is sometimes
cleaner. Python 3.10 introduced **structural pattern matching** with
the `match` statement, which fills the niche but is more capable than
a plain switch and not always the right tool for simple cases. Second,
the boolean keywords are `and`, `or`, `not`, not the C style `&&`,
`||`, `!`. Third, assignment is not allowed inside a regular `if`
condition the way it is in C; the **walrus operator** (`:=`,
introduced in 3.8) does that explicitly when really needed.

## 11. for loops

A `for` loop iterates over the elements of any iterable: a list, tuple,
string, dict, set, file, range, generator. There is no traditional C
style `for (i = 0; i < n; i++)`; Python's `for` is always over a
collection.

````
regions = ["Europe", "Americas", "Asia", "Pacific"]

for region in regions:
    print(region)
````

To loop a fixed number of times, use `range`. `range(n)` produces the
integers from 0 up to but not including `n`; `range(start, stop)` is
inclusive of `start` and exclusive of `stop`; `range(start, stop, step)`
takes a step.

````
for i in range(4):
    print(i)              # 0 1 2 3

for year in range(2025, 2028):
    print(year)           # 2025 2026 2027

for n in range(0, 100, 10):
    print(n)              # 0 10 20 ... 90
````

Two helpers turn the most common patterns into one liners. `enumerate`
yields `(index, value)` pairs, removing the need for a manual counter;
`zip` walks two or more iterables in lockstep.

````
for i, region in enumerate(regions):
    print(i, region)

revenue_values = [120_000, 250_000, 90_000, 60_000]
for region, amount in zip(regions, revenue_values):
    print(f"{region}: {amount:,}")
````

Looping over a dict yields its keys by default; `.items()` yields
`(key, value)` pairs, which is almost always what is actually wanted
inside the loop.

````
revenue = {"Europe": 120_000, "Americas": 250_000, "Asia": 90_000, "Pacific": 60_000}

for region, amount in revenue.items():
    print(region, amount)
````

A common newcomer instinct from C or Java is to loop with `for i in
range(len(regions)):` and then index `regions[i]`. Both `for region
in regions` and `for i, region in enumerate(regions)` read better and
are more idiomatic; the indexed form is rarely the right answer.

## 12. while loops, break, and continue

A `while` loop runs as long as its condition is truthy. It is the
right tool when the number of iterations is not known in advance:
reading from a file until end, polling a service until it answers,
processing a queue until empty.

````
remaining = 1000
withdrawals = 0

while remaining >= 100:
    remaining -= 100
    withdrawals += 1

print(withdrawals)        # 10
````

`break` exits the innermost loop immediately. `continue` skips the
rest of the current iteration and goes back to the loop header. Both
work in `for` loops and `while` loops alike.

````
revenue = [120_000, -1, 90_000, 60_000, -1, 30_000]
total = 0
for value in revenue:
    if value < 0:
        continue            # skip sentinel values
    if value > 200_000:
        break               # stop early on a large value
    total += value
````

Python adds a small twist: both `for` and `while` can have an `else`
clause that runs **only if the loop ran to completion without
hitting `break`**. This is unusual among mainstream languages and a
genuinely useful piece of syntax once internalised.

````
target = "Asia"
for region in regions:
    if region == target:
        print(f"found {target}")
        break
else:
    print(f"{target} not found")
````

The `else` is rarely the first thing reached for, but it is the right
tool when the loop is searching for something and a "did not find"
branch needs to follow. The alternative (a separate boolean flag) is
noisier.

For deliberate infinite loops, `while True:` plus a `break` inside is
the conventional form. Avoid it when the loop has a natural exit
condition that fits on the header line; reach for it when the exit
condition lives several lines deep inside the body.

## 13. Functions

A function is declared with `def`, takes zero or more parameters, and
optionally returns a value. The body is an indented block; the
function ends where the indentation does.

````
def total_revenue(revenue: dict[str, float]) -> float:
    total = 0.0
    for amount in revenue.values():
        total += amount
    return total

revenue = {"Europe": 120_000.0, "Americas": 250_000.0, "Asia": 90_000.0}
print(total_revenue(revenue))     # 460000.0
````

A function that has no `return` statement (or just a bare `return`)
implicitly returns `None`. The call site does not distinguish between
"explicitly returned `None`" and "fell off the end"; both produce the
same value.

Multiple return values are returned as a tuple, which the caller can
unpack:

````
def min_and_max(revenue: dict[str, float]) -> tuple[float, float]:
    values = list(revenue.values())
    return min(values), max(values)

low, high = min_and_max(revenue)
print(low, high)
````

Variables created inside a function are local to that function and
disappear when the function returns. Names defined at module level are
visible inside functions automatically (this is "lexical scope"); the
reverse is not true. To modify a module level variable from inside a
function, the `global` keyword is required, and the need for it is a
sign that the design should probably be different (pass the value in,
return the new value, or wrap the state in a class).

````
counter = 0

def increment_wrong():
    counter = counter + 1     # creates a local 'counter', then UnboundLocalError

def increment_right():
    global counter
    counter += 1
````

The Python community strongly prefers explicit arguments and return
values to global state; reach for `global` only as a last resort.

## 14. Default and keyword arguments

Function parameters can have default values, which makes the parameter
optional at the call site. Callers can also pass arguments by name
rather than by position, which makes long argument lists self
documenting.

````
def report(revenue: float, currency: str = "USD", precision: int = 2) -> str:
    return f"{revenue:,.{precision}f} {currency}"

report(1_234.5)                                # '1,234.50 USD'
report(1_234.5, currency="EUR")                # '1,234.50 EUR'
report(1_234.5, precision=0)                   # '1,235 USD'
report(1_234.5, precision=4, currency="GBP")   # '1,234.5000 GBP'
````

Two facts about defaults are worth pinning down on day one. First,
defaults are evaluated **once**, when the function is defined, not
each time it is called. For immutable defaults (numbers, strings,
`None`), this never causes trouble. For mutable defaults (lists,
dicts, sets) it causes the most surprising bug in beginner Python:

````
def append_region(region: str, regions: list[str] = []) -> list[str]:
    regions.append(region)
    return regions

append_region("Europe")     # ['Europe']
append_region("Asia")       # ['Europe', 'Asia']  the same list, persisted between calls
````

The fix is to use `None` as the default sentinel and create a fresh
list inside the function:

````
def append_region(region: str, regions: list[str] | None = None) -> list[str]:
    if regions is None:
        regions = []
    regions.append(region)
    return regions
````

Second, **keyword only parameters** can be enforced by placing them
after a bare `*` in the signature. This prevents callers from passing
those arguments positionally, which makes the call site readable when
the function has many parameters.

````
def write_report(revenue: float, *, currency: str = "USD", precision: int = 2) -> str:
    return f"{revenue:,.{precision}f} {currency}"

write_report(1_234.5, "EUR")           # TypeError: too many positional arguments
write_report(1_234.5, currency="EUR")  # OK
````

For functions with more than three or four parameters, keyword only
arguments are almost always the right choice. The call site reads as
prose, and adding a new parameter does not silently change the meaning
of existing calls.

## 15. Imports and the standard library

Code from outside the current file is brought in with `import`. The
imported module becomes a name in the current namespace, and its
contents are accessed through dot notation.

````
import math

print(math.sqrt(16))         # 4.0
print(math.pi)               # 3.141592653589793
````

The `from ... import ...` form imports specific names directly:

````
from math import sqrt, pi

print(sqrt(16))
print(pi)
````

Both forms are common. The plain `import math` form is preferred when
the module is used in many places, because the `math.sqrt` reference
makes the origin obvious. The `from math import sqrt` form is preferred
when only a few names are used and the file would otherwise be
cluttered with `math.` prefixes. The wildcard form (`from math import
*`) imports everything and is almost always the wrong choice; it
clutters the namespace and makes name origins invisible.

The `as` clause renames the imported name, which is conventional for a
few libraries:

````
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
````

The Python **standard library** ships with the interpreter and contains
a remarkable amount of useful code. A few modules every user reaches
for in the first few weeks:

- `math`: numeric functions (`sqrt`, `log`, `sin`, constants).
- `statistics`: mean, median, stdev, mode of a sample.
- `random`: pseudo random numbers, sampling, shuffling.
- `datetime`: dates, times, durations.
- `pathlib`: object oriented file path handling (`Path("data.csv").exists()`).
- `os` and `sys`: operating system and interpreter level access.
- `json`: encode and decode JSON to and from dicts and lists.
- `csv`: read and write CSV files.
- `collections`: `Counter`, `defaultdict`, `deque`, `namedtuple`.
- `itertools`: efficient iterators (`product`, `combinations`, `groupby`).
- `functools`: function tools (`partial`, `lru_cache`, `reduce`).

Reaching for `pip install some_thing` is rarely the first move. The
standard library covers more than first impressions suggest; a brief
look there before searching for a third party package saves time and
dependencies.

## 16. Reading and writing files

The built in function `open` returns a file object. The conventional
way to use it is the `with` block, which guarantees the file is closed
when the block exits, even if an exception is raised inside.

````
with open("sales.csv", "r", encoding="utf-8") as f:
    text = f.read()

print(text[:60])
````

The mode argument is a small string with one mode character and
optional flags. The common modes are `"r"` (read text), `"w"` (write
text, truncating any existing content), `"a"` (append text), and
adding `"b"` for binary (`"rb"`, `"wb"`). The `encoding=` argument is
worth supplying explicitly; the default depends on the platform and
quietly produces wrong characters on data that crosses systems.

For line by line reading, iterate the file directly. The lines come
back with their trailing newline characters; `rstrip` removes them.

````
with open("sales.csv", "r", encoding="utf-8") as f:
    for line in f:
        print(line.rstrip())
````

Writing flips the mode and uses `write` (which does not add a newline)
or `writelines` for an iterable of strings:

````
regions = ["Europe", "Americas", "Asia", "Pacific"]

with open("regions.txt", "w", encoding="utf-8") as f:
    for region in regions:
        f.write(region + "\n")
````

For tabular files, the `csv` module from the standard library handles
quoting, escaping, and line endings:

````
import csv

with open("sales.csv", "r", encoding="utf-8", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row["region"], row["revenue"])
````

For larger or richer tabular work the right tool is pandas (covered in
its own module). The plain `csv` module is the right tool when the
data is small, the schema is fixed, and the dependencies should stay
minimal.

## 17. Errors and tracebacks

When something goes wrong, Python raises an **exception** and prints a
**traceback**: a list of the function calls in progress at the moment
the error happened, ending with the exception type and message. The
traceback reads bottom up: the last line is what went wrong, the lines
above show how the program got there.

````
revenue = {"Europe": 120_000}
print(revenue["Mars"])
# Traceback (most recent call last):
#   File "report.py", line 2, in <module>
#     print(revenue["Mars"])
#           ~~~~~~~^^^^^^^^
# KeyError: 'Mars'
````

A handful of exception types come up daily.

- `NameError`: a variable name is not defined (typo, used before assignment).
- `TypeError`: a value of the wrong type was used (`"a" + 1`, calling something that is not callable).
- `ValueError`: a value of the right type but inappropriate content (`int("abc")`).
- `IndexError`: a sequence index is out of range.
- `KeyError`: a dict key is missing.
- `AttributeError`: a method or attribute does not exist on the object.
- `FileNotFoundError`: a file the program tried to open does not exist.
- `ZeroDivisionError`: division or modulo by zero.

When an error is expected and recoverable, wrap the risky call in
`try` / `except`:

````
def safe_divide(a: float, b: float) -> float:
    try:
        return a / b
    except ZeroDivisionError:
        return 0.0
````

The `except` clause can name a specific exception type, a tuple of
types, or be bare (catching everything). **Never catch a bare
`except:`** in regular code; it swallows `KeyboardInterrupt` and
makes errors invisible. Catch the specific exceptions that are
expected; let everything else propagate.

````
try:
    value = int(user_input)
except ValueError:
    print("not a number")
````

A `try` block can also have an `else` clause (runs if no exception was
raised) and a `finally` clause (runs no matter what, used for cleanup
that the `with` block does not cover).

The most common new user reflex is to catch and ignore exceptions to
"make the error go away". This is almost always the wrong move. The
exception is usually telling the truth, and silencing it leaves the
underlying problem unsolved. Read the traceback bottom up, fix the
cause, and keep the exception.

## 18. Comments, docstrings, and type hints

Comments start with `#` and run to the end of the line. They are
ignored by the interpreter and exist for human readers.

````
# total revenue across all regions
total = sum(revenue.values())
````

The convention is to comment **why**, not **what**. Code that says
`total = sum(revenue.values())` does not need a comment that says
"sum the values"; it does benefit from a comment explaining a
non obvious choice or constraint.

A **docstring** is the first statement of a function, class, or module
when that statement is a string literal. Tools (IDEs, the built in
`help` function, documentation generators) pick docstrings up
automatically. Triple quoted strings are the convention because
docstrings often span multiple lines.

````
def total_revenue(revenue: dict[str, float]) -> float:
    """Return the sum of revenue across all regions.

    Missing regions are treated as zero. Negative values are summed as is.
    """
    return sum(revenue.values())
````

Help on the function (`help(total_revenue)` in the REPL, or the same
text shown by an editor on hover) prints the docstring along with the
signature.

Python is dynamically typed, but it has supported optional **type
hints** since version 3.5. Hints have no runtime effect; they are read
by editors, type checkers (`mypy`, `pyright`), and human readers.

````
def report(revenue: dict[str, float], threshold: float = 100_000.0) -> list[str]:
    return [r for r, amount in revenue.items() if amount >= threshold]
````

The form is `name: type = default` for parameters, `-> type` for the
return value, and `name: type = value` for variables when the
inference would otherwise be ambiguous. The common types are the
obvious ones (`int`, `float`, `str`, `bool`), the parameterised
collection types (`list[int]`, `dict[str, float]`, `tuple[int, str]`),
optional types (`int | None` for "either an int or None"), and unions
(`int | str` for "either an int or a string").

Type hints are not required and the interpreter does not enforce them.
A program with hints runs identically to a program without them. The
value is in tooling: editors autocomplete more accurately, type
checkers catch a class of bugs that tests would otherwise have to
catch, and the function signature documents itself. The course uses
hints throughout starting from the next module; today's first program
will introduce them gently.

## 19. Putting it together: a first program

The pieces above combine into a recognisable end to end program. The
example below reads a small Sales Plan from a CSV, computes per region
totals and a comparison against a target, and prints a short report.
It uses every topic from this module: variables, strings, dicts,
control flow, functions with type hints, imports, file I/O, error
handling, and comments.

````
# sales_report.py
"""Read sales.csv, total by region, and compare against a target."""

from pathlib import Path
import csv

CSV_PATH: Path = Path("sales.csv")
TARGET:  float = 500_000.0


def read_sales(path: Path) -> dict[str, float]:
    """Load a CSV with columns 'region' and 'revenue' into a dict."""
    totals: dict[str, float] = {}
    with open(path, "r", encoding="utf-8", newline="") as f:
        reader = csv.DictReader(f)
        for row in reader:
            region = row["region"]
            revenue = float(row["revenue"])
            totals[region] = totals.get(region, 0.0) + revenue
    return totals


def format_line(region: str, revenue: float, target: float) -> str:
    share = revenue / target if target > 0 else 0.0
    flag = "OK " if revenue >= target else "    "
    return f"{flag} {region:<10} {revenue:>12,.2f}   ({share:.1%} of target)"


def main() -> None:
    try:
        totals = read_sales(CSV_PATH)
    except FileNotFoundError:
        print(f"No file at {CSV_PATH}")
        return

    if not totals:
        print("No data.")
        return

    grand_total = sum(totals.values())
    print(f"Sales report, target = {TARGET:,.0f}")
    print("-" * 50)
    for region in sorted(totals):
        print(format_line(region, totals[region], TARGET))
    print("-" * 50)
    print(f"   {'TOTAL':<10} {grand_total:>12,.2f}")


if __name__ == "__main__":
    main()
````

Two patterns in the program are worth naming explicitly. The
`if __name__ == "__main__":` guard at the bottom is the conventional
way to make a Python file usable both as a script (run directly,
`__name__` equals `"__main__"`) and as a module (imported from another
file, `__name__` equals the module name). The guard ensures `main()`
runs only when the file is executed, not when it is imported.

The second pattern is the layered structure: a small `read_sales`
function that does I/O, a small `format_line` function that does
formatting, and a `main` function that orchestrates them. This
separation is overkill for a thirty line program but is the habit that
keeps a three hundred line program readable. Each function does one
thing; `main` reads top to bottom.

Run it once with a small sample CSV:

```text
region,revenue
Europe,120000
Americas,250000
Asia,90000
Pacific,60000
Europe,15000
```

Output:

```text
Sales report, target = 500,000
--------------------------------------------------
     Americas       250,000.00   (50.0% of target)
     Asia            90,000.00   (18.0% of target)
     Europe         135,000.00   (27.0% of target)
     Pacific         60,000.00   (12.0% of target)
--------------------------------------------------
     TOTAL          535,000.00
```

This is a complete, working Python program. Everything in it has been
covered in the topics above; reading the source from top to bottom,
recognising which topic each line came from, is the goal.

## 20. Real world starter habits

Once the syntax stops being the obstacle, a few habits are worth
adopting from the first week. They are not advanced techniques; they
are the daily defaults that experienced Python users reach for without
thinking.

**Use a virtual environment.** Each project gets its own isolated set
of installed packages. The standard tool is `python -m venv .venv`,
which creates a `.venv/` folder; `source .venv/bin/activate` (or
`.venv\Scripts\activate` on Windows) makes that environment active.
Anything `pip install` adds is local to the project. The newer tools
`uv` and `poetry` automate the same workflow and are worth a look once
the basics are comfortable.

**Format the code automatically.** `black` (or its faster cousin
`ruff format`) reformats code to a consistent style with no manual
arguments to make. Style debates disappear; reviewing diffs gets
faster. Configure the editor to format on save and forget about it.

**Lint as part of the workflow.** `ruff` (or the older `flake8` and
`pylint`) catches a class of bugs and bad smells before the code
runs: unused imports, undefined names, shadowed builtins, suspicious
comparisons. Configure it to run on save or as a pre commit hook.

**Type check incrementally.** `mypy` or `pyright` consume the type
hints from Topic 18 and verify them. Starting strict on a brand new
project is reasonable; retrofitting strict typing onto an existing
project is best done a module at a time, with the strictness ramped up
as coverage grows.

**Prefer the standard library first.** Every dependency is a future
maintenance cost. For dates, paths, JSON, CSV, regex, math, basic
statistics, the standard library covers it. Reach for a package only
when the standard library is genuinely missing the feature, not as the
first move.

**Write small functions and small files.** A function that does one
thing fits in the head while reading; a function that does five things
needs to be re read on every call. The same applies to files: a
single file with twenty small functions is easier to navigate than one
file with two huge ones, and easier still than thirty files each with
one small function.

**Read the traceback bottom up.** The last line says what went wrong,
and the line above it shows where. Most beginner debugging time is
spent trying to fix the symptom several layers up the call stack.

**Commit often, in small steps.** Even on a one person project, git
is the safety net that turns "I broke something" into a thirty second
revert rather than an afternoon of reconstruction.

## 21. Common mistakes

A short collection of errors that come up often in first month Python,
each with the cleaner alternative.

**Indenting with mixed tabs and spaces.** Python treats tabs and
spaces as distinct, and a file that mixes them at the same logical
indentation level raises `IndentationError` or `TabError`. Configure
the editor to insert spaces (the convention is four per level) and
the issue disappears.

````
# Wrong: an invisible tab on the second line
def total():
    a = 1
	b = 2     # IndentationError on some files
    return a + b

# Correct: four spaces, every line
def total():
    a = 1
    b = 2
    return a + b
````

**Comparing to `None` with `==`.** The `is` operator is the
convention and what every linter expects. Most types behave the same
either way, but custom classes with overridden `__eq__` can produce
surprises.

````
# Wrong
if value == None:
    ...

# Correct
if value is None:
    ...
````

**Mutable default arguments.** Defaults are evaluated once at
definition time. A list or dict default is shared across every call
that does not pass the argument explicitly.

````
# Wrong: every call without 'items' shares one list
def add(item, items=[]):
    items.append(item)
    return items

# Correct
def add(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items
````

**Using `range(len(seq))` to iterate.** Iterating directly is shorter
and idiomatic; `enumerate` covers the case where the index is also
needed.

````
# Wrong
for i in range(len(regions)):
    print(regions[i])

# Correct
for region in regions:
    print(region)

# Correct (when the index is also wanted)
for i, region in enumerate(regions):
    print(i, region)
````

**Catching every exception.** A bare `except:` swallows
`KeyboardInterrupt`, `SystemExit`, and unrelated bugs. Catch the
specific exceptions that are expected.

````
# Wrong
try:
    do_thing()
except:
    pass

# Correct
try:
    do_thing()
except (ValueError, KeyError) as e:
    print(f"could not do thing: {e}")
````

**Modifying a list while iterating it.** The iterator and the list
share state; deleting items shifts the indices and skips elements.

````
values = [1, 2, 3, 4, 5]

# Wrong
for v in values:
    if v % 2 == 0:
        values.remove(v)
# values ends up [1, 3, 5] only by accident; for some inputs items are skipped

# Correct: build a new list
values = [v for v in values if v % 2 != 0]
````

**Forgetting that strings are immutable.** Methods that look like
they would mutate (`upper`, `replace`, `strip`) return a new string.
The original is unchanged.

````
text = "hello"

# Wrong: discards the result
text.upper()
print(text)            # 'hello'

# Correct
text = text.upper()
print(text)            # 'HELLO'
````

**Using `print` to debug and forgetting to remove the calls.**
Scattered `print` statements in production code are a smell, and they
mix with real output. The standard library `logging` module lets the
same calls be silenced, redirected, or filtered without code changes.

````
# Wrong: leaves debug output in production
def total(values):
    print("DEBUG: values =", values)
    return sum(values)

# Correct: use logging, control verbosity from outside
import logging
logger = logging.getLogger(__name__)

def total(values):
    logger.debug("values = %s", values)
    return sum(values)
````

**Importing with the wildcard form.** `from module import *` puts
every public name from the module into the current namespace,
clobbering existing names silently and making it impossible to tell
where a name came from on a later read.

````
# Wrong
from math import *
from statistics import *
mean(...)             # which module's mean?

# Correct
import statistics
import math
statistics.mean(...)
math.sqrt(...)
````

**Treating `is` as `==`.** `is` tests object identity; `==` tests
equality. They coincide for `None`, `True`, `False`, and small cached
integers, which makes the bug invisible until the values get bigger
or the types get more complex.

````
a = 1000
b = 1000

# Wrong: works for small ints because of caching, fails for larger ones
print(a is b)         # may print True or False depending on the value

# Correct
print(a == b)         # True
````
