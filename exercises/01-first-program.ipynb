Three pieces of Python show up in this exercise:

<details><summary>1. How is a list iterated in Python?</summary>

`for value in sales:` binds the name `value` to each entry of `sales` in turn. The indented body of the loop runs once per entry. The name `value` is a choice; any other name (such as `month_total` or `m`) works equally well, and Python does not treat any of them specially.

</details>

<details><summary>2. How are two numbers compared?</summary>

The comparison operators are `>` for "greater than", `<` for "less than", and `==` for "equal to". An expression like `value > target` evaluates to either `True` or `False`. These operators are the atoms that `if` statements branch on.

</details>

<details><summary>3. Which keywords build a three way decision?</summary>

`if`, then `elif`, then `else`. Python tests the conditions in order and runs the first branch whose condition evaluates to `True`. The remaining branches are skipped. The `else` clause carries no condition and runs when none of the earlier branches matched.

</details>

<details><summary>4. How is the equal case handled?</summary>

If the first branch tests `>` and the second branch tests `<`, then any value equal to the target falls through to the `else` clause naturally. There is no need for a third explicit `==` test. This is a small but pleasant property of three way comparisons against a single threshold.

</details>

<details><summary>5. How is text printed?</summary>

`print("over")` writes the string `over` followed by a newline. The argument to `print` can be any Python value, including numbers, lists, and strings. For this exercise a short label string is enough.

</details>

## Solution

The cell below holds a complete, working version of the exercise. It is offered as a reference rather than as the answer. The order of the comparisons and the choice of label strings are conventions, not the only correct version; an attempt that reaches the same output through a slightly different path is just as good.

The list `sales` is iterated once. For each value, two comparisons place it on one side of `target` or the other, and equality falls through to the `else` clause naturally, so no explicit `==` test is written. `print` writes the matching label. The loop variable is named `value` for readability.

A small habit worth picking up early: copy this code into a debugger (the VS Code Python debugger, or `pdb` from the terminal, both work) and step through it one line at a time. Watch `value` change on each iteration of the loop, and watch which branch of the `if`/`elif`/`else` runs for each value. The syntax becomes intuition only once the steps have been observed a few times in motion.

```
for value in sales:
    if value > target:
        print("over")
    elif value < target:
        print("under")
    else:
        print("on target")
```
