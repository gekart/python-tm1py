The approach is built from three functions and a single call. One function takes a list of floats and returns its arithmetic mean. A few pieces of Python show up here for the first time:

- `def name(parameter: type) -> return_type:` defines a function. Parameters are local names that hold the values passed in by the caller. The body is indented; `return value` hands a value back to the caller and ends the function. A function annotated `-> None` returns nothing meaningful and is used purely for its side effects (such as printing).
- `import statistics` makes the standard library `statistics` module available under the name `statistics`. The function `statistics.mean(values)` returns the arithmetic mean of any iterable of numbers, including a list of floats.
- A function defined at module level can read other module level names, including `TARGET` and `sales`, without those names being passed in as parameters. This is what people mean when they refer to a "global" inside a function. Reading is straightforward; rebinding a module level name from inside a function takes the `global` keyword and is a separate concern that this exercise does not cover.
- `def main() -> None:` is a convention. The work of the program lives inside `main`, and a single call to `main()` at the bottom of the cell runs it. In a script file the call is often guarded with `if __name__ == "__main__":`, which lets the file be imported without running the program; in a notebook a direct call is the simpler choice.

<details><summary>1. How is a function defined in Python?</summary>

`def function_name(parameter_name: type) -> return_type:` declares a function. The body is the indented block that follows. The function is not run when it is defined; it runs only when something calls it, e.g., `function_name(some_value)`. Parameter names are local to the function; they do not collide with names defined elsewhere.

</details>

<details><summary>2. How does a function return a value?</summary>

`return expression` hands the value of `expression` back to the caller and ends the function. A function that does not execute a `return` statement returns `None`. A function annotated `-> None` is one that exists for its side effects (such as printing) and returns nothing meaningful.

</details>

<details><summary>3. How is the `statistics` library used?</summary>

`import statistics` is written once, at the top of the file. After that, `statistics.mean(values)` returns the arithmetic mean of `values`, where `values` is any iterable of numbers, including a list. There is no need to write the formula by hand.

</details>

<details><summary>4. How can a function read a module level constant?</summary>

A function defined at module level can read any name defined at module level, including `TARGET`. Inside the function body, `TARGET` evaluates to the same value as the one set in the Data cell. Nothing extra is needed for reading. Rebinding a module level name from inside a function would require the `global` keyword, but this exercise only reads, not rebinds.

</details>

## Solution

The cell below holds a complete, working version of the exercise. As before, this is offered as a reference; the structure of the functions and the wording of the labels are conventions, not the only acceptable form.

The program reads `sales` and `TARGET` from the Data cell. `average` is a thin wrapper around `statistics.mean` that fixes the parameter type and the return type for clarity. `classify` reads `TARGET` directly from module level; it is the smallest example in this notebook of a function that uses a module level value rather than taking it as a parameter. `main` ties the two together: it computes and prints the year average, then iterates over `sales` to print each month alongside its classification. The very last line of the cell calls `main()` to run the program. In a Python script file the call is more often written as `if __name__ == "__main__": main()`, which lets the file be imported without running, but in a notebook the direct call is the simpler choice.

For a fuller feel of how the program runs, copy this code into a debugger (the VS Code Python debugger, or `pdb` from the terminal) and step through it one line at a time. Watching the call from `main` into `average` and back, then the loop and the calls into `classify`, makes the function abstraction concrete: the call stack, the parameter bindings, and the return values are all visible in the debugger window in a way they are not on the page.

```
import statistics

def average(values: list[float]) -> float:
    return statistics.mean(values)

def classify(value: float) -> str:
    if value > TARGET:
        return "over"
    elif value < TARGET:
        return "under"
    return "on target"

def main() -> None:
    avg = average(sales)
    print("Average:", avg)
    for value in sales:
        print(value, classify(value))

main()
```
